```python
import pandas as pd
```

```python
X = pd.read_csv('housing_X.csv')
y = pd.read_csv('housing_y.csv')

# to_csv -> Index=False
```

```python
# split to train and test
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=.2, random_state=42)
```

```python
ls.predict(X_test)
ls.coef_
```

```python

X = pd.read_csv('adult_X.csv').drop(columns='Unnamed: 0')
y = pd.read_csv('adult_y.csv').drop(columns='Unnamed: 0')
```

```python
X.columns
```

```python
X.dtypes
```

```python
# Pilih column sesuai type column

numeric = ['age','education-num','hours-per-week']
categorical = ['workclass','marital-status','occupation','native-country']
```

```python
# Pipeline

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# 1. Tahapan Preprocessing
ct = ColumnTransformer(transformers=[
    ('num', StandardScaler(), numeric), # encoding numeric menjadi standard scaler
    ('cat', OneHotEncoder(), categorical) # encoding categorical menjadi onehotencode
])
```

```python
from sklearn.pipeline import Pipeline

# Menyimpan proses-proses
pipeline_rfc = Pipeline([
    ('preprocessing', ct), # preprocessing
    ('model', best_rfc) # pemodelan
])
```

```python
# pisahkan datanya sebelum di fitting
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=.2, random_state=42)
```

```python
# Fitting
pipeline_rfc.fit(X_train,y_train)
```

```python
# Prediksi

pipeline_rfc.predict(X_test)
```

```python
# menyimpan pipeline
joblib.dump(pipeline_rfc, 'pipeline_rfc_joblib')
```

```python
# Membuat function
def prediksi():
    data = {}

    for i in numeric: # inputan numeric
        data[i] = float(input(f'masukkan nilai {i}:')) # str to float

    for i in categorical: # inputan categorical
            data[i] = input(f'masukkan nilai {i}:')

    data = pd.DataFrame(data, index=[0]) # convert ke DataFrame, harus ada index
    return pipeline_rfc.predict(data)
```

```python
# jalankan function prediksi
prediksi()
```
