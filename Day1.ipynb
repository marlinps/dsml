```python
# Data Structures

1. List [] -> mutable, ordered, allows duplicates
2. Tuple () -> immutable, ordered, allows duplicates
3. Set {} -> mutable, unordered, does not allow duplicates
4. Dictionary {} -> mutable, unordered, key-value pairs
```

```python
# list
list_buah = ['apel', 'jeruk', 'mangga', 'pisang', 'anggur']
```

```python
list_buah[0] # akses elemen pertama
list_buah[0:3] # akses elemen pertama, kedua, dan ketiga
```

```python
list_buah[::2] # stepping
```

```python
list_buah.append('kiwi') # menambahkan elemen di akhir list
```

```python
list_buah.pop()
```

```python
list_buah
```

```python
# tuple
tuple(list_buah) # konversi list ke tuple
```

```python
# Dictionary -> {key: value}
dict_buah = { 'buah' : ('apel', 'jeruk', 'mangga', 'pisang', 'anggur') }
print(dict_buah)
```

```python
# looping -> for each loop, while loop (digunakan ketika kita ingin mengulang selama suatu kondisi bernilai True)
i = 0
while i < 10:
    print(i)
    i += 1
```

```python
for i in range(10): # range(start, stop, step)
    print(i)
```

```python
for i in list_buah:
    print(i)
```

```python
for i in range(len(list_buah)):
    print(i)
```

```python
# list comprehension -> membuat list baru dari list yang sudah ada
[i for i in range(10)]
```

```python
[i for i in range(10) if i % 2 == 0]
```

```python
array = np.random.randint(1, 100, size=(3, 5))
```

```python
array
```

```python
array[array > 50] # array dengan elemen yang lebih besar dari 50
```

```python
array[(array > 50) & (array < 80)] # array dengan elemen yang lebih besar dari 50 dan lebih kecil dari 80
# and -> &, or -> |, not -> ~
```

```python
import seaborn as sns
```

```python
sns.get_dataset_names()
```

```python
df = sns.load_dataset('penguins')
```

```python
df.columns()
```

```python
df.head()
```

```python
df.tail()
```

```python
df.sample(60)
```

```python
df['species'] # Filter rows where species is 'Adelie'
```

```python
df[['species', 'island']] # Filter rows where species is 'Adelie'
```

```python
df[df['species'] == 'Adelie'] # Filter rows where species is 'Adelie'
```

```python
df[(df['island'] == 'Torgersen') & (df['bill_length_mm'] > 40)].reset_index(drop=True) # Filter rows where island is 'Torgersen' and bill_length_mm is greater than 40
```

```python
df.info()
```

```python
# loc -> label location/index
# iloc -> index location/index
```

```python
df.loc[6:10]
```

```python
df.iloc[6:10, 0:2] # Filter rows from index 6 to 9 and columns from index 0 to 2
```

```python
df.loc[6:10, ['species', 'island']] # Filter rows from index 6 to 9 and columns 'species' and 'island'
```
