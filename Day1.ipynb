```python
# Data Structures

1. List [] -> mutable, ordered, allows duplicates
2. Tuple () -> immutable, ordered, allows duplicates
3. Set {} -> mutable, unordered, does not allow duplicates
4. Dictionary {} -> mutable, unordered, key-value pairs
```

```python
# list
list_buah = ['apel', 'jeruk', 'mangga', 'pisang', 'anggur']
```

```python
list_buah[0] # akses elemen pertama
list_buah[0:3] # akses elemen pertama, kedua, dan ketiga
```

```python
list_buah[::2] # stepping
```

```python
list_buah.append('kiwi') # menambahkan elemen di akhir list
```

```python
list_buah.pop()
```

```python
list_buah
```

```python
# tuple
tuple(list_buah) # konversi list ke tuple
```

```python
# Dictionary -> {key: value}
dict_buah = { 'buah' : ('apel', 'jeruk', 'mangga', 'pisang', 'anggur') }
print(dict_buah)
```

```python
ktp = True
```

```python
# one linear expression
# result_if_true if condition else result_if_false

'boleh mengendari mobil' if ktp == True else 'tidak boleh mengendarai mobil'
```

```python
# looping -> for each loop, while loop (digunakan ketika kita ingin mengulang selama suatu kondisi bernilai True)
i = 0
while i < 10:
    print(i)
    i += 1
```

```python
for i in range(10): # range(start, stop, step)
    print(i)
```

```python
for i in list_buah:
    print(i)
```

```python
for i in range(len(list_buah)):
    print(i)
```
