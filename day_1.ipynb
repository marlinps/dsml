```python
df['color'] = 'hitam'
```

```python
df['species'].unique() # menampilkan nilai unik pada kolom 'species' pada
# Adelie -> hitam, Chinstrap -> cokelat, Gentoo -> abu-abu
```

```python
def warna(x):
    if x == 'Adelie':
        return 'hitam'
    elif x == 'Chinstrap':
        return 'cokelat'
    else:
        return 'abu-abu'
```

```python
# for i in range(len(df)):
#     warna(df['species'].values[i])
```

```python
# apply -> function untuk menerapkan fungsi ke setiap elemen dalam kolom atau baris DataFrame
df['color'] = df['species'].apply(warna)
```

```python
# def warna(x):
#     if x == 'Adelie':
#        print('hitam')
#     elif x == 'Chinstrap':
#         print('cokelat')
#     else:
#         print('abu-abu')

# one linear expression  and nameless function
df['color'] = df['species'].apply(lambda x: 'hitam' if x == 'Adelie' else 'cokelat' if x == 'Chinstrap' else 'abu-abu')
```

```python
df
```

```python
# bill depth > 15 dan body_mass > 3000 -> besar 
df['size'] = df.apply(lambda x: 'besar' if (x['bill_depth_mm'] > 15) & (x['body_mass_g'] > 3000) else 'kecil', axis=1)
```

```python
df['size'].unique() # menampilkan nilai unik pada kolom 'size' pada DataFrame
```

```python
len(df[df['size'] == 'besar'])
```

```python
# menghapus column
df.drop('color', axis=1, inplace=True)

# df.drop(columns='color', inplace=True)
```
